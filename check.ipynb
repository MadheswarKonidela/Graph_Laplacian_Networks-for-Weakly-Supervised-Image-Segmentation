```python
import os
import numpy as np
from PIL import Image

def check_submission(folder_path):
    errors = []
    
    # 1. Check if there are any subfolders
    subfolders = [f for f in os.listdir(folder_path) if os.path.isdir(os.path.join(folder_path, f))]
    if subfolders:
        errors.append(f"❌ Found subfolders: {subfolders}. All images must be directly inside the folder.")

    # 2 & 3. Check each image
    for fname in os.listdir(folder_path):
        fpath = os.path.join(folder_path, fname)
        if not os.path.isfile(fpath):
            continue
        try:
            img = Image.open(fpath)
            arr = np.array(img).astype(np.uint8)
            print(np.unique(arr))

            # --- (2) Check dimensions ---
            if arr.shape[1] != 500 or arr.shape[0] != 375:
                errors.append(f"❌ {fname}: Wrong size {arr.shape[1]}x{arr.shape[0]}, expected 500x375.")

            # --- (3) Check pixel values ---
            unique_vals = np.unique(arr)
            if not set(unique_vals).issubset({0,1}):
                errors.append(f"❌ {fname}: Invalid pixel values {unique_vals}, expected only 0 and 1.")

        except Exception as e:
            errors.append(f"❌ {fname}: Could not open/read ({e})")

    if errors:
        print("\n".join(errors))
    else:
        print("✅ All checks passed! Submission is valid.")


check_submission("/Users/madheswarkonidela/Downloads/Submission_1_-_7075740_Madheswar_Konidela")
```

```python
import zipfile
import numpy as np
from PIL import Image
import io

def check_submission_zip(zip_path):
    errors = []

    with zipfile.ZipFile(zip_path, 'r') as zf:
        file_list = zf.namelist()

        # --- (1) Check for nested folders ---
        for f in file_list:
            if f.endswith("/"):
                errors.append(f"❌ Found folder inside zip: {f}. All images must be directly inside the zip.")
            elif "/" in f:
                errors.append(f"❌ Found nested path '{f}'. Files must be at top level.")

        # --- (2 & 3) Check each image ---
        for f in file_list:
            if f.endswith("/") or not (f.lower().endswith(".png") or f.lower().endswith(".jpg")):
                continue
            try:
                with zf.open(f) as file:
                    img = Image.open(io.BytesIO(file.read()))
                    arr = np.array(img).astype(np.uint8)

                    # Check dimensions
                    if arr.shape[1] != 500 or arr.shape[0] != 375:
                        errors.append(f"❌ {f}: Wrong size {arr.shape[1]}x{arr.shape[0]}, expected 500x375.")

                    # Check pixel values
                    unique_vals = np.unique(arr)
                    if not set(unique_vals).issubset({0,1}):
                        errors.append(f"❌ {f}: Invalid pixel values {unique_vals}, expected only 0 and 1.")

            except Exception as e:
                errors.append(f"❌ {f}: Could not open/read ({e})")

    if errors:
        print("\n".join(errors))
    else:
        print("✅ All checks passed! Submission is valid.")


check_submission_zip("/Users/madheswarkonidela/Downloads/Submission_1_-_7075740_Madheswar_Konidela.zip")
```
